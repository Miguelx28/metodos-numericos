# Resumen de Implementación

## Atributos privados

```python
_name
_health
_mana
_level
```

## Propiedades

```python

def name(self):
```

```python

```

```python
@health.setter
def health(self, value):
```

```python
@property

```

```python

def mana(self, value):
```

```python
@property
def level(self):
```

## Métodos

```python
def level_up(self):
```

```python
def __str__(self):
```

````python
class GameCharacter:

    def __init__(self, name):
        self._name = name
        self.health = 100
        self.mana  = 50
        self._level = 1

    @property
    def name(self):
        return self._name

    @property
    def health(self):
        return self._health

    @health.setter
    def health(self, salud):
        if salud < 0:
            self._health = 0
        elif  salud > 100:
            self._health = 100
        else:
            self._health = salud
    
    @property
    def mana(self):
        return self._mana

    @mana.setter
    def mana(self, mana):
        if mana < 0:
            self._mana = 0
        elif mana > 50: 
            self._mana = 50
        else:
            self._mana = mana

    @property
    def level(self):
        return self._level

    def level_up(self):
        self._level += 1
        self.health = 100
        self.mana = 50
        print(f"{self._name} leveled up to {self._level}")

    def __str__(self):
        return f"Name: {self._name}\nLevel: {self._level}\nHealth: {self.health}\nMana: {self.mana}"


hero = GameCharacter('Kratos')
hero.health = -10
hero.health = 75
hero.mana = 100
print(hero)
print(hero.health)
print(hero.mana)
hero.level_up()
````
