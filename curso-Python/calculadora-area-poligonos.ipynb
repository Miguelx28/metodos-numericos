# Desarrolla una Calculadora de Área de Polígonos

En este proyecto, usarás programación orientada a objetos para crear una clase `Rectangle` y una clase `Square`. La clase `Square` debe ser una subclase de `Rectangle` y heredar sus métodos y atributos.

## Objetivo

Cumplir con las historias de usuario que se presentan a continuación y pasar todas las pruebas para completar el laboratorio.

## Historias de usuario

1. Debes crear una clase `Rectangle`.

2. Cuando se crea un objeto `Rectangle`, debe inicializarse con los atributos `width` y `height`.

   La clase también debe contener los siguientes métodos:

   - `set_width`: establece el ancho del rectángulo.
   - `set_height`: establece la altura del rectángulo.
   - `get_area`: devuelve el área:

     ```text
     ancho × alto
     ```

   - `get_perimeter`: devuelve el perímetro:

     ```text
     2 × (ancho + alto)
     ```

   - `get_diagonal`: devuelve la diagonal:

     ```text
     √(ancho² + alto²)
     ```

   - `get_picture`: devuelve una cadena que representa la figura usando líneas de `*`.

     El número de líneas debe ser igual a la altura y el número de caracteres `*` de cada línea debe ser igual al ancho.

     Debe haber un carácter de nueva línea (`\n`) al final de cada línea.

     Si el ancho o la altura es mayor que `50`, debe devolver la cadena:

     ```text
     Too big for picture.
     ```

   - `get_amount_inside`: acepta otra figura, ya sea un cuadrado o un rectángulo, como argumento.

     Devuelve el número de veces que la figura pasada como argumento cabe dentro de la figura que invoca el método, sin permitir rotaciones.

     Por ejemplo, un rectángulo con un ancho de `4` y una altura de `8` puede contener dos cuadrados con lados de `4`.

3. Si una instancia de `Rectangle` se representa como una cadena, debe verse de la siguiente manera:

   ```text
   Rectangle(width=5, height=10)
   ```

4. Debes crear una clase `Square` que sea una subclase de `Rectangle`.

5. Cuando se crea un objeto `Square`, debe inicializarse con una sola longitud de lado.

   El método `__init__` debe almacenar la longitud del lado en los atributos `width` y `height` heredados de la clase `Rectangle`.

6. La clase `Square` debe contener los siguientes métodos:

   - `set_width`: sobrescribe el método `set_width` de la clase `Rectangle`. Debe establecer el ancho y la altura con la longitud recibida.
   - `set_height`: sobrescribe el método `set_height` de la clase `Rectangle`. Debe establecer el ancho y la altura con la longitud recibida.
   - `set_side`: establece el ancho y la altura del cuadrado con la longitud del lado.

7. La clase `Square` debe poder acceder a los métodos de la clase `Rectangle`.

8. Si una instancia de `Square` se representa como una cadena, debe verse de la siguiente manera:

   ```text
   Square(side=9)
   ```

## Pruebas

1. Debe existir una clase `Rectangle`.

2. Debe existir una clase `Square`.

3. La clase `Square` debe ser una subclase de la clase `Rectangle`.

4. La clase `Square` debe ser una clase diferente de la clase `Rectangle`.

15. Una instancia de la clase `Square` debe tener una representación en cadena diferente después de establecer nuevos valores mediante:

    ```python

    ```

16. Una instancia de la clase `Square` debe tener una representación en cadena diferente después de establecer nuevos valores mediante:

    ```python
    set_width()
    ```

    o:

    ```python
    set_height()
    ```

17. El método `get_picture()` debe devolver una representación en cadena de una instancia de `Rectangle`.

18. El método `get_picture()` debe devolver una representación en cadena de una instancia de `Square`.

19. El método `get_picture()` debe devolver:

    ```text
    Too big for picture.
    ```

    si el atributo `width` o el atributo `height` es mayor que `50`.

20. La siguiente expresión debe devolver `6`:

    ```python
    Rectangle(15, 10).get_amount_inside(Square(5))
    ```

21. La siguiente expresión debe devolver `1`:

    ```python
    Rectangle(4, 8).get_amount_inside(Rectangle(3, 6))
    ```

22. La siguiente expresión debe devolver `0`:

    ```python
    Rectangle(2, 3).get_amount_inside(Rectangle(3, 6))
    ```

````python
import math
class Rectangle:

    def __init__(self, width: float, height: float) -> None:
        self.width = width
        self.height = height


    def set_width(self, width: float) -> None:
        self.width =  width
    
    def set_height(self, height: float) -> None:
        self.height = height
    
    def get_area(self) -> float:
        return self.height * self.width
    
    def get_perimeter(self) -> float:
        return 2 * (self.width + self.height )
    
    def get_diagonal(self) -> float:
        return math.sqrt((self.height**2) + (self.width**2))
    
    def get_picture(self) -> str:
        picture = ''
        if self.height > 50 or self.width >50:
            return 'Too big for picture.'
        for i in range(self.height): 
            picture += '*'* self.width + '\n'
        return picture

    def get_amount_inside(self, figure) -> int:
        return int((self.get_area() / figure.get_area()))

    def __str__(self) -> str:
        return f'Rectangle(width={self.width}, height={self.height})'


class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)

    def set_height(self, side):
        self.height =  side
        self.width = side

    def set_width(self, side):
        self.width = side
        self.height = side

    def set_side(self, side):
        self.width  = side
        self.height = side

    def __str__(self):
        return f'Square(side={self.height})'



rect = Rectangle(10, 5)
print(rect.get_area())

rect.set_height(3)
print(rect.get_perimeter())
print(rect)
print(rect.get_picture())

sq = Square(9)
print(sq.get_area())

sq.set_side(4)
print(sq.get_diagonal())
print(sq)
print(sq.get_picture())

rect.set_height(8)
rect.set_width(16)
print(rect.get_amount_inside(sq))
        
````
