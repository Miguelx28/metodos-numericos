# Construye un Generador de Rango de Números

En este laboratorio, construirás una función `range_of_numbers` que utiliza recursión para generar una lista de números dentro de un rango especificado.

Por ejemplo:

- `range_of_numbers(3, 9)` debe devolver:

  ```python
  [3, 4, 5, 6, 7, 8, 9]
  ```

- `range_of_numbers(5, 5)` debe devolver:

  ```python
  [5]
  ```

## Historias de usuario

1. Debes crear una función llamada `range_of_numbers` con dos parámetros llamados `start_num` y `end_num`.

2. La función debe devolver una lista de enteros consecutivos que comienza con `start_num` y termina con `end_num`, incluyendo ambos valores.

3. El valor de `start_num` siempre será menor o igual que `end_num`.

4. Tu función debe usar recursión llamándose a sí misma.

5. Tu función no debe usar:

   - Ciclos `for`
   - Ciclos `while`
   - Comprensiones de listas
   - La función `range()`

6. Cuando `start_num` sea igual a `end_num`, la función debe devolver una lista que contenga `start_num`.

   Este es el **caso base**.

7. Para el caso recursivo, la función debe llamarse a sí misma con un argumento que la acerque al caso base y luego agregar el número actual a la lista devuelta.

8. La función no debe usar variables globales para almacenar la lista devuelta.

# Pruebas

1. Debes tener una función llamada `range_of_numbers` con los parámetros `start_num` y `end_num`.

2. `range_of_numbers(1, 5)` debe devolver:

   ```python
   [1, 2, 3, 4, 5]
   ```

3. `range_of_numbers(6, 9)` debe devolver:

   ```python
   [6, 7, 8, 9]
   ```

4. `range_of_numbers(4, 4)` debe devolver:

   ```python
   [4]
   ```

5. `range_of_numbers(10, 15)` debe devolver:

   ```python
   [10, 11, 12, 13, 14, 15]
   ```

6. La función `range_of_numbers` debe llamarse a sí misma.

7. La función `range_of_numbers` no debe usar ciclos, comprensiones ni la función `range()`.

8. Llamar a:

   ```python
   range_of_numbers(1, 3)
   ```

   no debe afectar llamadas posteriores.

   Después de esa llamada:

   ```python
   range_of_numbers(6, 9)
   ```

   debe seguir devolviendo:

   ```python
   [6, 7, 8, 9]
   ```

````python
def range_of_numbers(start_num, end_num):
    if start_num > end_num:
        return []
    numbers  = range_of_numbers(start_num, end_num-1)
    numbers.append(end_num)
    return numbers

range_of_numbers(6,9)
````
