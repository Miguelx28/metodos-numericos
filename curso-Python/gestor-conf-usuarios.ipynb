````python
def convertir_tupla_min(tupla):
    return tuple(elemento.lower() for elemento in tupla)

def add_setting(diccionario, tupla):
    #validar diccionario -> diccionario y tupla -> tupla
    if not isinstance(diccionario, dict):
        return "Error: El primer argumento debe ser un diccionario."
    if not isinstance(tupla, tuple):
        return "Error: El segundo argumento debe ser una tupla."
    tupla = convertir_tupla_min(tupla)
    #verificar si key exite en diccionerio
    if tupla[0] in diccionario:
        return f"Setting '{tupla[0]}' already exists! Cannot add a new setting with this name."
    diccionario[tupla[0]] = tupla[1]
    return f"Setting '{tupla[0]}' added with value '{tupla[1]}' successfully!"


def update_setting(diccionario, tupla):
    tupla = convertir_tupla_min(tupla)
    if tupla[0] in diccionario:
        diccionario.update({tupla[0]: tupla[1]})
        return f"Setting '{tupla[0]}' updated to '{tupla[1]}' successfully!"
    return f"Setting '{tupla[0]}' does not exist! Cannot update a non-existing setting."

def delete_setting(diccionario, llave):
    llave = llave.lower()
    if llave in diccionario:
        del diccionario[llave]
        return f"Setting '{llave}' deleted successfully!"
    return f"Setting not found!"

def view_settings(diccionario):
    if not diccionario:
        return "No settings available."

    resultado = "Current User Settings:\n"

    for clave, valor in diccionario.items():
        resultado += f"{clave.capitalize()}: {valor}\n"

    return resultado




####Test 

test_settings = {
    'theme': 'dark',
    'notifications':'enable',
    'Volume':'high' 
}


````
