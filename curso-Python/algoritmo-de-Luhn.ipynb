# Implementar el algoritmo de Luhn

El algoritmo de Luhn, también conocido como el algoritmo "módulo 10" o "mod 10", es una fórmula de suma de verificación simple utilizada para validar una variedad de números de identificación, como los números de tarjetas de crédito.

Estos son los pasos para validar un número usando el algoritmo de Luhn:

1. Comenzando desde la derecha, y excluyendo el dígito más a la derecha (el dígito de verificación), duplica el valor de cada otro dígito.
2. Si el resultado de duplicar un dígito es mayor que `9`, suma los dígitos para obtener un solo dígito. Alternativamente, puedes restar `9` del resultado.
3. Toma la suma de todos los dígitos, incluyendo el dígito de verificación.
4. Si la suma de todos los dígitos es un múltiplo de `10`, entonces el número es válido; de lo contrario, no es válido.

Por ejemplo, considera el número `453914881`. Los pasos para validarlo usando el algoritmo de Luhn son:

```text
Account number      4   5   3   9   1   4   8   8   1
Double every other  4  10   3  18   1   8   8  16   1
Sum 2-char digits   4   1   3   9   1   8   8   7   1
```

1. Luego suma todos los números:

   ```text
   4 + 1 + 3 + 9 + 1 + 8 + 8 + 7 + 1 = 42
   ```

2. Dado que `42` no es un múltiplo de `10`, el número es inválido.

## Ejemplos

## Pruebas

1. Deberías tener una función llamada `verify_card_number`.
2. `verify_card_number('453914889')` debería devolver `VALID!`.
3. `verify_card_number('4111-1111-1111-1111')` debería devolver `VALID!`.
4. `verify_card_number('453914881')` debería devolver `INVALID!`.
5. `verify_card_number('1234 5678 9012 3456')` debería devolver `INVALID!`.
6. `verify_card_number` debería devolver `VALID!` cuando se llame con un número de tarjeta de crédito válido.
7. `verify_card_number` debería devolver `INVALID!` cuando se llame con un número de tarjeta de crédito inválido.

````python
def verify_card_number(card_number):
    # Eliminar guiones y espacios
    card_number = card_number.replace('-', '').replace(' ', '')

    total = 0

    # Recorrer los dígitos de derecha a izquierda
    for index, digit in enumerate(reversed(card_number)):
        digit = int(digit)

        # Duplicar cada segundo dígito, excluyendo el dígito de verificación
        if index % 2 == 1:
            digit *= 2

            if digit > 9:
                digit -= 9

        total += digit

    # Comprobar si la suma es múltiplo de 10
    if total % 10 == 0:
        return 'VALID!'

    return 'INVALID!'


print(verify_card_number('453914889'))
print(verify_card_number('4111-1111-1111-1111'))
print(verify_card_number('453914881'))
print(verify_card_number('1234 5678 9012 3456'))
````
