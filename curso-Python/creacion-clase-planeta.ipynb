## Pruebas

1. Debes crear una clase llamada `Planet`.

2. La clase `Planet` debe tener un método `__init__`.

3. El método `__init__` debe tener cuatro parámetros:

   ```text
   self, name, planet_type, star
   ```

   en ese orden.

4. El método `__init__` debe lanzar un `TypeError` con el mensaje:

   ```text
   name, planet type, and star must be strings
   ```

   si el primer argumento pasado no es de tipo string.

5. El método `__init__` debe lanzar un `TypeError` con el mensaje:

   ```text
   name, planet type, and star must be strings
   ```

   si el segundo argumento pasado no es de tipo string.

6. El método `__init__` debe lanzar un `TypeError` con el mensaje:

   ```text
   name, planet type, and star must be strings
   ```

   si el tercer argumento pasado no es de tipo string.

7. El método `__init__` debe lanzar un `ValueError` con el mensaje:

   ```text
   name, planet_type, and star must be non-empty strings
   ```

   si el primer argumento pasado es una cadena vacía.

8. El método `__init__` debe lanzar un `ValueError` con el mensaje:

   ```text
   name, planet_type, and star must be non-empty strings
   ```

   si el segundo argumento pasado es una cadena vacía.

9. El método `__init__` debe lanzar un `ValueError` con el mensaje:

   ```text
   name, planet_type, and star must be non-empty strings
   ```

   si el tercer argumento pasado es una cadena vacía.

10. Debes asignar `name` a `self.name` en el método `__init__`.

11. Debes asignar `planet_type` a `self.planet_type` en el método `__init__`.

12. Debes asignar `star` a `self.star` en el método `__init__`.

13. La clase `Planet` debe tener un método `orbit`.

14. El método `orbit` debe devolver una cadena con el formato:

    ```text
    {name} is orbiting around {star}...
    ```

15. La clase `Planet` debe tener un método `__str__`.

16. El método `__str__` debe devolver una cadena con el formato:

    ```text
    Planet: {name} | Type: {planet_type} | Star: {star}
    ```

17. Debes crear tres instancias de la clase `Planet` llamadas:

    - `planet_1`
    - `planet_2`
    - `planet_3`

18. Debes imprimir cada objeto planeta para ver la salida del método `__str__`.

19. Debes llamar al método `orbit` en cada objeto planeta e imprimir el resultado.

````python
class Planet:
    def __init__(self, name, planet_type, star):

        if not isinstance(name,str):
            raise TypeError('name, planet type, and star must be strings')
        elif not isinstance(planet_type,str):
            raise TypeError('name, planet type, and star must be strings')
        elif not isinstance(star, str):
            raise TypeError('name, planet type, and star must be strings')
        elif name == '':
            raise ValueError('name, planet_type, and star must be non-empty strings')
        elif planet_type == '':
            raise ValueError('name, planet_type, and star must be non-empty strings')
        elif star == '':
            raise ValueError('name, planet_type, and star must be non-empty strings')
        self.name = name
        self.planet_type = planet_type
        self.star = star

    def orbit(self):
        return f'{self.name} is orbiting around {self.star}...'

    def __str__(self):
        return f'Planet: {self.name} | Type: {self.planet_type} | Star: {self.star}'

planet_1 = Planet('tierra', 'tierra', 'luna')
planet_2 = Planet('tierra1', 'tierra2', 'luna1')
planet_3 = Planet('tierra2', 'tierra1', 'luna2')


print(planet_1)
print(planet_2)
print(planet_3)

print(planet_1.orbit())
print(planet_2.orbit())
print(planet_3.orbit())
````
