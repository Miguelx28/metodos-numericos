# Construir una Tabla Hash

En este laboratorio, construirás una tabla hash desde cero. Una tabla hash es una estructura de datos que almacena pares clave-valor. Una tabla hash funciona tomando la clave como entrada y luego aplicándole una función hash específica.

Para este laboratorio, la función hash será simple: sumará los valores Unicode de cada carácter de la clave. El valor hash se utilizará como la clave real para almacenar el valor asociado. El mismo valor hash también se utilizará para recuperar y eliminar el valor asociado con la clave.

## Objetivo

Cumplir con las historias de usuario a continuación y pasar todas las pruebas para completar el laboratorio.

## Historias de usuario

1. Debes definir una clase llamada `HashTable` con un atributo `collection` inicializado como un diccionario vacío cuando se crea una nueva instancia de `HashTable`.

   El diccionario `collection` debe almacenar pares clave-valor basados en el valor hash de la clave.

2. La clase `HashTable` debe tener cuatro métodos de instancia:

   - `hash`
   - `add`
   - `remove`
   - `lookup`

3. El método `hash` debe:

   - Tomar una cadena como parámetro.
   - Devolver un valor hash calculado como la suma de los valores Unicode (ASCII) de cada carácter de la cadena.
   - Puedes usar la función `ord` para realizar el cálculo.

4. El método `add` debe:

   - Aceptar una clave y un valor como argumentos.
   - Calcular el hash de la clave.
   - Usar el valor hash calculado como clave para almacenar un diccionario que contenga el par clave-valor dentro de `collection`.
   - Si varias claves producen el mismo valor hash, sus pares clave-valor deben almacenarse en el diccionario anidado existente bajo el mismo valor hash.

5. El método `remove` debe:

   - Tomar una clave como argumento.
   - Calcular su hash.
   - Confirmar si la clave existe en `collection`.
   - Eliminar el par clave-valor correspondiente de la tabla hash.
   - Si la clave no existe en la colección, no debe lanzar errores ni eliminar nada.

6. El método `lookup` debe:

   - Tomar una clave como argumento.
   - Calcular el hash de la clave.
   - Devolver el valor correspondiente almacenado en la tabla hash.
   - Si la clave no existe en `collection`, debe devolver `None`.

22. Cuando agregas dos claves que producen el mismo hash, como `fcc` y `cfc`, la colección debe verse así:

````python
class HashTable:
    def __init__(self):
        self.collection = {}

    def hash(self, string):
        total = 0
        for char in string:
            total += ord(char)
        return total
        

    def add(self, key, value):
        hash_key = self.hash(key)
        if hash_key not in self.collection:
            self.collection[hash_key] = {}
#        self.collection[hash_key] = value
        self.collection[hash_key][key] = value
        

    def remove(self, key):
        hash_key = self.hash(key)
        if not hash_key in self.collection:
            return
        if not key in self.collection[hash_key]:
            return  
        del self.collection[hash_key][key]
        
    
    def lookup(self, arg):
        hash_key = self.hash(arg)
        if not hash_key in  self.collection:
            return None
        return self.collection[hash_key].get(arg)
        
````
