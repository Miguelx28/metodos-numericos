# Implementar el algoritmo de la Torre de Hanoi

En este laboratorio, resolverás el rompecabezas matemático conocido como la **Torre de Hanoi**. El rompecabezas consiste en tres varillas y un número de discos de diferentes diámetros.

El rompecabezas comienza con los discos apilados en la primera varilla, en tamaño decreciente, con el disco más pequeño arriba y el disco más grande abajo.

El objetivo del rompecabezas Torre de Hanoi es mover todos los discos a la última varilla. Para hacerlo, debes seguir tres reglas simples:

1. Puedes mover solo los discos que estén arriba.
2. Puedes mover solo un disco a la vez.
3. No puedes colocar discos más grandes encima de discos más pequeños.

8. `hanoi_solver(n)` debería resolver el rompecabezas de la Torre de Hanoi para cualquier valor positivo de `n`.

````python
def hanoi_solver(n):
    rods = [
        list(range(n, 0, -1)),
        [],
        []
    ]

    movements = [f'{rods[0]} {rods[1]} {rods[2]}']

    def move_disks(number_of_disks, source, auxiliary, destination):
        if number_of_disks == 0:
            return

        # Mover n - 1 discos de la varilla inicial a la auxiliar
        move_disks(
            number_of_disks - 1,
            source,
            destination,
            auxiliary
        )

        # Mover el disco superior a la varilla de destino
        disk = rods[source].pop()
        rods[destination].append(disk)

        # Guardar el estado actual de las tres varillas
        movements.append(f'{rods[0]} {rods[1]} {rods[2]}')

        # Mover n - 1 discos de la varilla auxiliar al destino
        move_disks(
            number_of_disks - 1,
            auxiliary,
            source,
            destination
        )

    move_disks(n, 0, 1, 2)

    return '\n'.join(movements)


print(hanoi_solver(3))
````
