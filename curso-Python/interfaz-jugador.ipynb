## Pruebas

- Esperando: Debes tener una clase llamada `Player`.
- Esperando: La clase `Player` debe heredar de la clase `ABC` del módulo `abc`.
- Esperando: La clase `Player` debe tener un método `__init__`.
- Esperando: El método `__init__` de `Player` debe tener un solo parámetro `self`.
- Esperando: El método `__init__` de `Player` debe establecer `self.moves` como una lista vacía.
- Esperando: El método `__init__` de `Player` debe establecer `self.position` en `(0, 0)`.
- Esperando: El método `__init__` de `Player` debe establecer `self.path` como una lista que contenga la posición inicial.
- Esperando: La clase `Player` debe tener un método `make_move`.
- Esperando: El método `make_move` de `Player` debe tener un solo parámetro `self`.
- Esperando: El método `make_move` de `Player` debe usar `random.choice` para obtener un movimiento aleatorio del atributo `moves`.
- Esperando: El método `make_move` de `Player` debe actualizar el atributo `position` sumando las coordenadas del movimiento seleccionado aleatoriamente.
- Esperando: El método `make_move` de `Player` debe agregar la nueva tupla `position` al atributo `path`.
- Esperando: El método `make_move` de `Player` debe devolver el atributo `position` actualizado.
- Esperando: La clase `Player` debe tener un método `level_up`.
- Esperando: El método `level_up` de `Player` debe tener un solo parámetro `self`.
- Esperando: El método `level_up` de `Player` debe ser un método abstracto.
- Esperando: La clase `Player` debe ser una clase abstracta.
- Esperando: Debes tener una clase llamada `Pawn`.
- Esperando: La clase `Pawn` debe heredar de la clase `Player`.
- Esperando: La clase `Pawn` debe tener un método `__init__`.
- Esperando: El método `__init__` de `Pawn` debe tener un solo parámetro `self`.
- Esperando: El método `__init__` de `Pawn` debe llamar al `__init__` del padre usando la función `super`.
- Esperando: El método `__init__` de `Pawn` debe establecer el atributo `moves` como una lista de tuplas que representan coordenadas `x, y`, donde cada tupla representa un movimiento de `1` unidad hacia arriba, abajo, izquierda o derecha.
- Esperando: La clase `Pawn` debe tener un método `level_up`.
- Esperando: El método `level_up` de `Pawn` debe agregar los cuatro movimientos diagonales de `1` unidad al atributo `moves`.

```python
from abc import ABC, abstractmethod
import random
class Player(ABC):
    def __init__(self) -> None:
        super().__init__()
        self.moves = []
        self.position = (0,0)
        self.path = [self.position]

    def make_move(self) -> tuple:
        move = random.choice(self.moves)
        self.position = (
            move[0] + self.position[0],
            move[1] + self.position[1]
        )
        self.path.append(self.position)
        return self.position

    @abstractmethod
    def level_up(self):
        pass
    

class Pawn(Player):
    def __init__(self) -> None:
        super().__init__()
        self.moves = [
            (0, 1), #arriba
            (0,-1), #abajo
            (-1,0), #izquierda
            (1, 0), #derecha 
        ]
    
    def level_up(self):
        diagonales = [
            (1, 1), #arriba+derecha
            (1,-1), #abajo+derecha
            (-1,1), #arriba+izquierda
            (-1,-1) #abajo+izquierda 
        ]
        for move in diagonales:
            self.moves.append(move)
```
