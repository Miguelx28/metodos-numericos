# Implementar el algoritmo de ordenación por selección

La ordenación por selección (Selection Sort) es otro algoritmo de ordenación popular que se enseña en la mayoría de los cursos de ciencias de la computación.

Este algoritmo funciona buscando repetidamente el elemento más pequeño de la parte no ordenada de la lista e intercambiándolo con el primer elemento no ordenado. Comienza seleccionando el valor mínimo de toda la lista y cambiándolo con el primer elemento. Luego pasa a la segunda posición, encuentra el valor más pequeño entre los elementos restantes sin ordenar y lo intercambia con el segundo elemento. Este proceso continúa hasta que toda la lista queda ordenada.

La ordenación por selección tiene una complejidad temporal cuadrática en el mejor, promedio y peor caso: `O(n²)`. La complejidad espacial es constante, `O(1)`, porque el ordenamiento se realiza en el mismo arreglo y utiliza una cantidad fija de memoria independientemente del tamaño de la lista.

````python
def selection_sort(arr):
    n = len(arr)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr
````
