## Historias de usuario

1. Debes definir una función llamada `quick_sort` para implementar el algoritmo Quicksort.

2. La función `quick_sort` debe recibir una lista de enteros como parámetro y devolver una nueva lista con los elementos ordenados de menor a mayor.

3. Para implementar el algoritmo debes:

   - Elegir un valor pivote de los elementos de la lista de entrada (usa el primer o el último elemento de la lista).
   - Dividir la lista de entrada en tres sublistas:
     - una con los elementos menores que el pivote,
     - una con los elementos iguales al pivote,
     - una con los elementos mayores que el pivote.
   - Llamar recursivamente a `quick_sort` para ordenar las sublistas y concatenarlas para producir la lista final ordenada.

## Pruebas

1. Debes tener una función llamada `quick_sort`.

2. La función `quick_sort` debe aceptar un único parámetro.

3. `quick_sort([])` debe devolver una lista vacía.

4. La función `quick_sort` no debe modificar la lista recibida como argumento.

5. `quick_sort([20, 3, 14, 1, 5])` debe devolver:

   ```python

   ```

6. `quick_sort([83, 4, 24, 2])` debe devolver:


7. `quick_sort([4, 42, 16, 23, 15, 8])` debe devolver:


8. `quick_sort([87, 11, 23, 18, 18, 23, 11, 56, 87, 56])` debe devolver:


9. No debes importar módulos ni usar métodos de ordenación integrados.

````python
def quick_sort(array):
    if len(array) <1:
        return array
    pivote = array[0]
    eq = [x for x in array if x == pivote]
    high = [x for x in array if x > pivote]
    less = [x for x in array if x < pivote]

    return quick_sort(less) + eq + quick_sort(high)
````
