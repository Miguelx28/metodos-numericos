# Implementa el Método de Bisección

El método de bisección, también conocido como método de búsqueda binaria, utiliza una búsqueda binaria para encontrar las raíces de una función de valor real.

Funciona reduciendo progresivamente el intervalo en el que se encuentra la raíz cuadrada hasta que el resultado está dentro de una tolerancia especificada.

Por ejemplo, si la tolerancia es `0.01`, el método de bisección continuará dividiendo el intervalo por la mitad hasta que la diferencia entre los límites inferior y superior sea menor o igual que `0.01`.

En este laboratorio, implementarás una función que utiliza el método de bisección para encontrar la raíz cuadrada de un número.

````python

def square_root_bisection(n, tol=1e-10, iter=10):
    if n < 0:
        raise ValueError('Square root of negative number is not defined in real numbers')
    elif n == 0 or n == 1:
        print(f'The square root of {n} is {n}')
        return n
    low = 0
    high  = max(1, n)
    for x in range(iter):
        x = (low + high)/2
        if ( high-low <= tol):
            print(f'The square root of {n} is approximately {x}')
            return x
        if x * x  < n:
            low = x
        else:
            high = x
    print(f'Failed to converge within {iter} iterations')
    return None
         

square_root_bisection(81, 1e-3, 50)
# The square root of 81 is approximately 9.0

square_root_bisection(225, 1e-7, 100)
# The square root of 225 is approximately 15.0

square_root_bisection(225, 1e-7, 10)
# Failed to converge within 10 iterations
square_root_bisection(0.001, 1e-7,50)
    
````
