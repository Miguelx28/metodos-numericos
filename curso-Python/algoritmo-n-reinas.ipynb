# Implementa el Algoritmo de las N-Reinas

El problema de las N-Reinas te pide colocar N reinas en un tablero de ajedrez de N×N de manera que ninguna de las dos reinas se ataque entre sí (ninguna comparta fila, columna o diagonal).

Por ejemplo, si hay un tablero 4x4, una estructura válida es:

```python
[1, 3, 0, 2]
```

Eso significa que en la fila 0, la reina está colocada en la columna 1; en la fila 1, la reina está colocada en la columna 3; en la fila 2, la reina está colocada en la columna 0; y en la fila 3, la reina está colocada en la columna 2.

Visualmente, esta estructura se ve así:

```text
. Q . .
. . . Q
Q . . .
. . Q .
```

Donde `Q` representa una reina y `.` representa una casilla vacía.

En este laboratorio, implementarás el solucionador del problema de las N-Reinas usando el enfoque de búsqueda en profundidad.

**Objetivo:** Cumple las historias de usuario a continuación y logra que todas las pruebas pasen para completar el laboratorio.

## Historias de usuario

1. Debes tener una función llamada `dfs_n_queens`.

2. La función debe aceptar exactamente un argumento: un entero `n`.

3. Si `n` es menor que `1`, la función debe devolver una lista vacía (`[]`).

4. La función debe devolver una lista de soluciones; cada solución es a su vez una lista de longitud `n`, donde el elemento en el índice `i` es el índice de columna (basado en 0) de la reina en la fila `i`.

## Pruebas

1. Debes tener una función llamada `dfs_n_queens` que tome un argumento.

2. Si `n` es menor que `1`, la función debe devolver una lista vacía.

3. La función debe devolver una lista de soluciones, donde cada solución es una lista de longitud `n`.

4. `dfs_n_queens(1)` debería devolver:

```python
[[0]]
```

5. `dfs_n_queens(2)` debería devolver:

```python
[]
```

6. `dfs_n_queens(3)` debería devolver:

```python
[]
```

7. `dfs_n_queens(4)` debería devolver:

```python
[[1, 3, 0, 2], [2, 0, 3, 1]]
```

8. `dfs_n_queens(5)` debería devolver:

```python
[
    [0, 2, 4, 1, 3],
    [0, 3, 1, 4, 2],
    [1, 3, 0, 2, 4],
    [1, 4, 2, 0, 3],
    [2, 0, 3, 1, 4],
    [2, 4, 1, 3, 0],
    [3, 0, 2, 4, 1],
    [3, 1, 4, 2, 0],
    [4, 1, 3, 0, 2],
    [4, 2, 0, 3, 1]
]
```

9. `len(dfs_n_queens(5))` debería ser:

```python
10
```

10. `len(dfs_n_queens(8))` debería ser:

```python
92
```

11. `dfs_n_queens` debe devolver el resultado correcto.

````python
def dfs_n_queens(n: int):
    if n < 1:
        return []
    
    soluciones = []
    pila = [[]]

    while pila:
        actual = pila.pop()
        fila = len(actual)

        if fila == n:
            soluciones.append(actual)
            continue

        for columna in range(n-1, -1,-1):
            posicion_valida = True

            for fila_ant, columna_ant in enumerate(actual):
                misma_columna = columna == columna_ant
                misma_diagonal = abs(fila-fila_ant) == abs(columna -columna_ant)
                if misma_columna or misma_diagonal:
                    posicion_valida = False
                    break
            if posicion_valida:
                pila.append(actual + [columna])

    return soluciones


print(dfs_n_queens(1))
# [[0]]

print(dfs_n_queens(2))
# []

print(dfs_n_queens(3))
# []

print(dfs_n_queens(4))
# [[1, 3, 0, 2], [2, 0, 3, 1]]

print(len(dfs_n_queens(5)))
# 10

print(len(dfs_n_queens(8)))
# 92
````
