```python
import numpy as np
import matplotlib.pyplot as plt

def taylor_ln(x, n):
    """Polinomio de Taylor de grado n de ln(x) alrededor de x0 = 1."""
    suma = 0
    for k in range(1, n + 1):
        suma = suma + ((-1) ** (k + 1)) * (x - 1) ** k / k
    return suma
```

```python
taylor_ln(1.5, 3)
```

### Gráfica 1
En una misma figura, $ln x$ y $P_1,P_2,P_3, y P_0$ en [0.1, 3], con n.linespace(0.1, 3, 400)
Limite el eje vertical a [-3,2] (plt.ylim), marque el punto(1.0) y agregue leyenda y nombres de los ejes

```python
x = np.linspace(0.1, 3, 400)

plt.figure(figsize=(8, 5))
plt.plot(x, np.log(x), 'k', lw=2, label=r'$\ln x$')
for n in [1, 2, 3, 6]:
    plt.plot(x, taylor_ln(x, n), label=rf'$P_{n}(x)$')
plt.plot(1, 0, 'ro', label=r'$(1,0)$')
plt.ylim(-3, 2)
plt.xlabel('x')
plt.ylabel('y')
plt.title(r'$\ln x$ y sus polinomios de Taylor en $x_0=1$')
plt.legend()
plt.grid(True)
plt.show()
```

```python
ns = np.arange(1, 31)
err15 = [abs(np.log(1.5) - taylor_ln(1.5, n)) for n in ns]
err25 = [abs(np.log(2.5) - taylor_ln(2.5, n)) for n in ns]
cota  = 0.5**(ns + 1) / (ns + 1)

plt.figure(figsize=(8, 5))
plt.semilogy(ns, err15, 'o-', label=r'Error en $x=1.5$')
plt.semilogy(ns, err25, 's-', label=r'Error en $x=2.5$')
plt.semilogy(ns, cota, 'k--', label=r'Cota $0.5^{n+1}/(n+1)$ en $x=1.5$')
plt.xlabel('n')
plt.ylabel(r'$|\ln x - P_n(x)|$')
plt.title('Error de aproximación')
plt.legend()
plt.grid(True, which='both')
plt.show()
```
